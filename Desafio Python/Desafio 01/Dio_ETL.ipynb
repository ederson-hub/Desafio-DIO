```python
# Extract
import pandas as pd

df = pd.read_csv('/content/SDW2025.csv', encoding='latin1', sep=';')

print("Top 5 rows of the DataFrame:")
print(df.head())

```

```python
# Transform
salario = {
    'Financeiro': 5000,
    'Marketing': 4500,
    'Estagiario': 1500,
    'Recepcionista': 2000,
    'Almoxerifado': 2500,
    'Administrativo': 3000,
    'Gerente de Vendas': 8000,
    'Gerente de TI': 8000,
    'Supervisor de produção': 4000,
    'CEO': 15000
}

df['Salario'] = df['Cargo'].map(salario).fillna(df['Salario'])

print("DataFrame update 'Salario' :")
print(df.head())
```

```python
# Load
df.to_json('sdw2025.json', orient='records', indent=4)
print("DataFrame successfully saved to 'sdw2025.json'")
```
